Segundo a Organização Mundial da Saúde (OMS), atualmente, mais pessoas morrem anualmente por essas enfermidades do que por qualquer outra causa. Essas doenças são de caráter sistêmico, o que implica no comprometimento de outros órgãos, em função da circulação sanguínea, como fonte de oxigênio e glicose.

Diversos fatores de risco estão associados ao desenvolvimento de DCV, os quais podem ser modificáveis ou não modificáveis. Os fatores modificáveis incluem hiperlipidemia, tabagismo, etilismo, obesidade, sedentarismo, má alimentação, etc; já os não modificáveis incluem fatores hereditários, idade, sexo e raça.


Desta forma, estima-se que cerca de 90% dessas doenças poderiam ser evitadas realizando pequenas mudanças na dieta e estilo de vida, é possível melhorar a saúde do coração e a qualidade de vida.

O que são Doenças Cardiovasculares (DCV)?
As doenças do coração são alterações no funcionamento do sistema cardíaco, sendo este responsável por transportar oxigênio e nutrientes necessários às células.

De acordo com a OMS, as doenças do coração incluem

* Doença coronariana;
* Doença cerebrovasculas;
* Doença arterial periférica;
* Doença cardíaca reumática;
* Cardiopatia congênita;
* Doença reumática; entre outras.

Antes de iniciar a análise dos dados, precisamos compreender o significado de cada variável que compõe o *dataset*. Assim, abaixo, podemos visualizar a lista com a descrição de todos os atributos.

**Informações sobre os atributos:**


* `age` - idade em anos
* `sex` - sexo do paciente
  * 0: mulher
  * 1: homem
* `cp` - tipo da dor torácica
  * 1: angina típica
  * 2: angina atípica
  * 3: dor não cardíaca
  * 4: assintomática
* `trestbps` - pressão arterial em repouso
* `chol` - colesterol sérico (*mg/dl*)
* `fbs` - açucar no sangue em jejum > 120*mg/dl*
  * 0: `False`
  * 1: `True`
* `restecg` - resultado do eletrocardiografia de repouso
  * 0: normal
  * 1: anormalidades de ST-T  (inversão da onda T e elevação ou depressão de > 0.05*mV*)
  * 2: hipertrofia ventricular esquerda provável ou definitiva (pelos critérios de Romhilt-Estes)
* `thalach` - frequência cardíaca máxima atingida
* `exang` - angina induzida pelo exercício
  * 0: não
  * 1: sim
* `oldpeak` - depessão do segmento ST induzida pelo exercício em relação ao repouso
* `slope` - inclinação do segmento ST no pico do exercício
* `ca` - número de vasos principais colorido por fluoroscopia
* `thal` - *teste de esforço cardíaco*
  * 3: normal
  * 6: defeito fixo
  * 7: defeito reversível
* `num` - diagnóstico de doença cardíaca (diagnóstico angiográfico)
  * 0: não há doença cardíaca ( < 50% de estreitamento do diâmetro)
  * 1,2,3,4: há doença cardíaca ( > 50% de estreitamento do diâmetro)

Para as análises realizadas neste projeto, a principal variável de interesse será `num`, uma vez que este atributo nos fornece informações se o paciente possui ou não doenças cardiovasculares.

Observando as informações sobre os atributos, percebemos que nossa variável alvo (`num`) possui cinco valores possíveis: \[0, 1, 2, 3, 4\].

No entanto, seguindo a maioria das pesquisas que utilizaram esse *dataset*, iremos considerar apenas duas situações possíveis:

* **0** - não há doença cardíaca
* **1** - há doença cardíaca

Então, iremos igualar a 1 os valores compreendidos entre 1-4.

```python
# VARIÁVEL ALVO ANTES

df_uci.num.value_counts()
```

```python
# VARIÁVEL ALVO DEPOIS

df_uci.num.value_counts()
```

# **Aplicação do modelo de Machine Learning**

Nesta etapa do projeto, implementaremos um modelo de Machine Learning para prever os casos em que um determinado paciente possui doença cardiovascular ou não. Assim, é necessário que o modelo de Machine Learning dê uma classificação ao ser alimentado com diversas variáveis independentes (*features*).

Para a criação do modelo de Machine Learning, utilizaremos a biblioteca ***Scikit-Learn***, que possui vários algoritmos de classificação, regressão e agrupamento incluindo máquinas de vetores de suporte, florestas aleatórias, gradient boosting, k-means e DBSCAN, e é projetada para interagir com as bibliotecas Python numéricas e científicas NumPy e SciPy.

### **Preparação dos Dados**

**Nesta etapa do projeto serão utilizados os dados do *dataset* df_uci**

Em todos os projetos de Machine Learning a etapa de pré-processamento de dados é fundamental. Aplicando as técnicas corretas para preparação dos dados, podemos aumentar consideravelmente o desempenho dos modelos implementados.

Para alimentar o modelo de Machine Learning, é necessário separar a variável alvo (dependente) do *dataset* das demais (independentes).

Geralmente, as variáveis são divididas como X e y, este projeto vai adotar o mesmo padrão.

Após separar nossas variáveis devidamente, como parte do pré-processamento dos dados, que irão alimentar o modelo de Machine Learning, será usado o **StardardScaler**, que vem junto com **sklearn.preprocessing**, para padronizar nossos dados numéricos.

```python
# IMPORTAÇÃO DAS FERRAMENTAS PARA O PRÉ-PROCESSAMENTO DOS DADOS

from sklearn.preprocessing import StandardScaler
```

```python
# SEPARAÇÃO DAS VARIÁEIS

X = df_uci.drop('num', axis = 1)
y = df_uci.num
```

```python
# PADRONIZAÇÃO DOS DADOS NUMÉRICOS

X = StandardScaler().fit_transform(X)
```

Por fim, antes de implementar o modelo, iremos dividir o nosso *dataset* em dados de treino e de teste.

Essa etapa também é fundamental, uma vez que ela evita que o modelo possa fazer previsões erradas ao "viciar", em palavras simples, o modelo com um determinado conjunto de dados. Ou seja, quando nós usamos os mesmos dados para treinar e testar nossos modelos, é provável que ele não aprenda a classificar os dados, mas apenas "decorar" as respostas. Assim, ao submetir esse modelo para novos dados, ele não vai conseguir uma boa acurácia.

Por isso, é fundamental usar dados diferentes para treino e teste.

```python
# DIVISÃO DOS DADOS EM TREINO E TESTE

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.28)
```

### **Modelo Random Forest**

Random Forest (floresta aleatória) é um algoritmo de aprendizagem supervisionada. Como você pode perceber pelo seu nome, ele cria uma floresta de um modo aleatório.

Segundo Niklas Donges, em um artigo do Towards Data Science, de modo simples: o algoritmo de florestas aleatórias cria várias árvores de decisão e as combina para obter uma predição com maior acurácia e mais estável.

<center><img alt = "Analisando Airbnb" width = "15%" src="https://static.thenounproject.com/png/961660-200.png"></center>

Uma grande vantagem do algoritmo de florestas aleatórias é que ele pode ser utilizado tanto para tarefas de classificação quanto para regressão, o que representa a maioria dos sistemas de aprendizagem de máquina atuais.

Você pode conferir um artigo completo do **Towards Data Science** sobre o modelo Random Forest clicando [aqui](https://towardsdatascience.com/understanding-random-forest-58381e0602d2).

```python
# IMPORTANDO O MODELO RANDOM FOREST

from sklearn.ensemble import RandomForestClassifier
```

```python
# INSTANCIANDO O MODELO

RF_model = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 42)
```

```python
# TREINANDO O MODELO

RF_model.fit(X_train, y_train)
```

### **Verificando o desempenho do modelo**

Uma das principais métricas utilizadas para avaliar o desempenho de um modelo de Machine Learning é a acurácia. Contudo, é importante ressaltar que cada caso é particular, com isso, uma boa acurácia nem sempre garantirá que o modelo está bom.

No caso da detecção de doenças cardiovasculares, o interesse é maximizar o número de verdadeiros positivos e minimizar o número de verdadeiros positivos e minimizar o número de falsos negativos, apesar de ser uma tarefa difícil de realizar.

Neste projeto, além da métrica de acurácia será utilizada a função classification_report para ver ver o desempenho do modelo sobre as métricas de precisão, ***recall, f1-score e support***.

```python
# IMPORTAÇÃO DAS FERRAMENTAS DE MÉTRICAS

from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
```

```python
# REALIZANDO A PREVISÃO NOS DADOS DE TESTE

y_pred = RF_model.predict(X_test)
```

```python
# VIZUALIZANDO A ACURÁCIA GERAL

print('[Acurácia] Random Forest:', accuracy_score(y_test, y_pred))
```

```python
# VISUALIZANDO O CLASSIFICATION REPORT

print('\n[Classification Report] Random Forest')

print(classification_report(y_test, y_pred))
```

**Descrição das métricas**

* **Precision** - Precisão de previsões positivas..
* **Reacall** - Fração de positivos que foram identificados corretamente.
* **f1-score** - representa a média harmonica entre precision e recall, sendo 0 o pior valor e 1 o melhor.

Aparentemente, o modelo apresenta bons resultados nas métricas avaliadas. Contudo, como citado anteriormente, isso não pode garantir totalmente que o modelo seja bom.

Para complementar a nossa análise de desempenho, é interessante visualizar a **matriz de confusão**

<center><img alt = "Analisando Airbnb" width = "50%" src="https://miro.medium.com/max/938/1*s7VB26Cfo1LdVZcLou-e0g.png"></center>

* **VP** - Se o paciente tiver doença cardíaca e o modelo prever que possui.
* **FP** - Se o paciente não tiver doença cardíaca e o modelo prever que possui.
* **FN** - Se o paciente tiver doença cardíaca e o modelo prever que não.
* **VN** - Se o paciente não tiver doença cardíaca e o modelo prever que não.

```python
# VISUALIZANDO A MATRIZ DE CONFUSÃO

pd.DataFrame(confusion_matrix(y_test, y_pred),
             index=['neg', 'pos'], columns=['pred_neg', 'pred_pos'])
```

A matriz de confusão funciona da seguinte maneira:

Nas linhas constam os casos reais e nas colunas, as previsões feitas pelo modelo.

Por exemplo, existem 35 casos realmente negativos e o modelo previu corretamente. Contudo, em 7 casos também negativos, ele previu como positivo.

    <h2 style="margin: 0; color: #0f172a; font-size: 20px;">1. Estatisticas Basicas de Atributos Clinicos Complementares</h2>
</div>
<p style="font-family: 'Segoe UI', Arial, sans-serif; color: #334155; font-size: 14px; line-height: 1.6; margin-top: 10px;">
    Na primeira etapa da exploracao, a analise e visual concentrou-se nas variaveis <code>num</code>, <code>sex</code>, <code>age</code> e <code>chol</code>. 
    Para ampliar o entendimento clinico da base <code>df_uci</code>, selecionamos quatro atributos fundamentais ainda nao explorados:
</p>
<ul style="font-family: 'Segoe UI', Arial, sans-serif; color: #334155; font-size: 14px; line-height: 1.6;">
    <li><strong><code>trestbps</code>:</strong> Pressao arterial em repouso na admissao hospitalar (mm Hg).</li>
    <li><strong><code>thalach</code>:</strong> Frequencia cardiaca maxima alcancada durante o teste de esforco (bpm).</li>
    <li><strong><code>oldpeak</code>:</strong> Depressao do segmento ST induzida pelo exercicio em relacao ao repouso.</li>
    <li><strong><code>cp</code>:</strong> Tipo de dor toracica (1: angina tipica, 2: angina atipica, 3: dor nao cardiaca, 4: assintomatica).</li>
</ul>

```python
# Estatisticas descritivas gerais dos atributos numericos ineditos
atributos_ineditos_num = ['trestbps', 'thalach', 'oldpeak']

stats_gerais = df_uci[atributos_ineditos_num].agg(
    ['count', 'mean', 'median', 'std', 'min', 'max']
).T.round(2)
stats_gerais.columns = ['Contagem', 'Media', 'Mediana', 'Desvio_Padrao', 'Minimo', 'Maximo']

print('=== Estatisticas Gerais dos Atributos Numericos Ineditos ===')
display(stats_gerais)

# Estatisticas estratificadas pela variavel-alvo (0 = Sem Doenca, 1 = Com Doenca)
stats_por_classe = df_uci.groupby('num')[atributos_ineditos_num].agg(['count', 'mean', 'median', 'min', 'max']).round(2)
stats_por_classe.index = ['Sem Doenca (0)', 'Com Doenca (1)']

print('\n=== Estatisticas Estratificadas por Diagnostico (num) ===')
display(stats_por_classe.T)

# Analise estatistica do atributo categorico inedito: cp (Tipo de Dor Toracica)
mapa_cp = {
    1: '1 - Angina Tipica',
    2: '2 - Angina Atipica',
    3: '3 - Dor Nao Cardiaca',
    4: '4 - Assintomatica'
}
cp_resumo = df_uci.groupby('cp')['num'].agg(
    Contagem_Total='count',
    Casos_Positivos='sum',
    Taxa_Doenca_Percentual=lambda s: round(s.mean() * 100, 2)
)
cp_resumo['Percentual_Base (%)'] = (cp_resumo['Contagem_Total'] / len(df_uci) * 100).round(2)
cp_resumo.index = cp_resumo.index.map(mapa_cp)

print('\n=== Distribuicao e Incidencia de Doenca por Tipo de Dor Toracica (cp) ===')
display(cp_resumo[['Contagem_Total', 'Percentual_Base (%)', 'Casos_Positivos', 'Taxa_Doenca_Percentual']])
```

<div style="border-bottom: 2px solid #2563eb; padding-bottom: 6px; margin-top: 24px; font-family: 'Segoe UI', Arial, sans-serif;">
    <h2 style="margin: 0; color: #0f172a; font-size: 20px;">2. Novos Graficos Baseados no Fluxograma de Escolha de Visualizacoes</h2>
</div>
<p style="font-family: 'Segoe UI', Arial, sans-serif; color: #334155; font-size: 14px; line-height: 1.6; margin-top: 10px;">
    Seguindo as recomendacoes do fluxograma <em>choosing-a-good-chart-09.pdf</em> (<em>Chart Suggestions - A Thought-Starter</em>, de Andrew Abela), cada visualizacao abaixo foi escolhida de acordo com o objetivo analitico e a natureza das variaveis:
</p>
<ul style="font-family: 'Segoe UI', Arial, sans-serif; color: #334155; font-size: 14px; line-height: 1.6;">
    <li><strong>Grafico 1 (Distribuicao entre grupos - Boxplot):</strong> Para comparar a distribuicao de variaveis numericas continuas (<code>thalach</code> e <code>oldpeak</code>) entre dois grupos distintos (<code>Sem Doenca</code> vs. <code>Com Doenca</code>), o fluxograma indica graficos de distribuicao como histogramas ou boxplots comparativos.</li>
    <li><strong>Grafico 2 (Comparacao entre categorias - Barras Agrupadas):</strong> Para comparar contagens entre categorias discretas de dor toracica (<code>cp</code>) segmentadas pelo diagnostico (<code>num</code>), o ramo <em>Comparison &rarr; Among Items</em> recomenda o grafico de barras verticais/agrupadas.</li>
    <li><strong>Grafico 3 (Relacao entre duas variaveis - Dispersao com Segmentacao):</strong> Para investigar a relacao conjunta entre <code>thalach</code> e <code>oldpeak</code> na presenca ou ausencia da doenca, o ramo <em>Relationship</em> recomenda o grafico de dispersao (<em>Scatter Chart</em>).</li>
</ul>

```python
# Configuracao de rotulos legiveis para plotagem
df_plot = df_uci.copy()
df_plot['Diagnostico'] = df_plot['num'].map({0: 'Sem Doenca (0)', 1: 'Com Doenca (1)'})
df_plot['Tipo_Dor'] = df_plot['cp'].map({
    1: 'Angina Tipica',
    2: 'Angina Atipica',
    3: 'Dor Nao Cardiaca',
    4: 'Assintomatica'
})

paleta_diag = {'Sem Doenca (0)': '#2563eb', 'Com Doenca (1)': '#dc2626'}

# GRAFICO 1: Boxplots comparativos (thalach e oldpeak por Diagnostico)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df_plot, x='Diagnostico', y='thalach', hue='Diagnostico', palette=paleta_diag, legend=False, ax=axes[0], width=0.5)
axes[0].set_title('Distribuicao da Frequencia Cardiaca Maxima (thalach)', fontsize=12, fontweight='bold', color='#0f172a')
axes[0].set_xlabel('Diagnostico', fontsize=10)
axes[0].set_ylabel('Frequencia Cardiaca Maxima (bpm)', fontsize=10)

sns.boxplot(data=df_plot, x='Diagnostico', y='oldpeak', hue='Diagnostico', palette=paleta_diag, legend=False, ax=axes[1], width=0.5)
axes[1].set_title('Distribuicao da Depressao do Segmento ST (oldpeak)', fontsize=12, fontweight='bold', color='#0f172a')
axes[1].set_xlabel('Diagnostico', fontsize=10)
axes[1].set_ylabel('Depressao ST (oldpeak)', fontsize=10)

plt.tight_layout()
plt.show()

# GRAFICO 2: Barras agrupadas de Tipo de Dor Toracica (cp) por Diagnostico
fig, ax = plt.subplots(figsize=(10, 5))
ordem_cp = ['Angina Tipica', 'Angina Atipica', 'Dor Nao Cardiaca', 'Assintomatica']
sns.countplot(data=df_plot, x='Tipo_Dor', hue='Diagnostico', order=ordem_cp, palette=paleta_diag, ax=ax)

ax.set_title('Ocorrencia de Doenca Cardiovascular por Tipo de Dor Toracica (cp)', fontsize=13, fontweight='bold', color='#0f172a')
ax.set_xlabel('Tipo de Dor Toracica', fontsize=11)
ax.set_ylabel('Quantidade de Pacientes', fontsize=11)
ax.legend(title='Diagnostico')

# Rotulos numericos nas barras
for container in ax.containers:
    ax.bar_label(container, padding=3, fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

# GRAFICO 3: Dispersao entre thalach e oldpeak segmentada por Diagnostico
fig, ax = plt.subplots(figsize=(10, 5))
sns.scatterplot(
    data=df_plot,
    x='thalach',
    y='oldpeak',
    hue='Diagnostico',
    palette=paleta_diag,
    alpha=0.75,
    s=55,
    ax=ax
)
ax.set_title('Relacao entre Frequencia Cardiaca Maxima (thalach) e Depressao ST (oldpeak)', fontsize=12, fontweight='bold', color='#0f172a')
ax.set_xlabel('Frequencia Cardiaca Maxima (bpm)', fontsize=11)
ax.set_ylabel('Depressao do Segmento ST (oldpeak)', fontsize=11)
ax.legend(title='Diagnostico')

plt.tight_layout()
plt.show()
```

<div style="border-bottom: 2px solid #2563eb; padding-bottom: 6px; margin-top: 24px; font-family: 'Segoe UI', Arial, sans-serif;">
    <h2 style="margin: 0; color: #0f172a; font-size: 20px;">3. Treinamento e Avaliacao de Novos Classificadores</h2>
</div>
<p style="font-family: 'Segoe UI', Arial, sans-serif; color: #334155; font-size: 14px; line-height: 1.6; margin-top: 10px;">
    Alem da abordagem inicial com <code>RandomForestClassifier</code>, treinamos e comparamos <strong>tres classificadores distintos</strong> utilizando a biblioteca <code>scikit-learn</code>:
</p>
<ol style="font-family: 'Segoe UI', Arial, sans-serif; color: #334155; font-size: 14px; line-height: 1.6;">
    <li><strong>Regressao Logistica (<code>LogisticRegression</code>):</strong> Modelo linear probabilistico amplamente adotado em triagem clinica pela interpretabilidade dos coeficientes.</li>
    <li><strong>Maquina de Vetores de Suporte (<code>SVC</code> com kernel RBF):</strong> Classificador baseado na maximizacao de margens de separacao entre classes em espacos nao lineares.</li>
    <li><strong>Gradient Boosting (<code>GradientBoostingClassifier</code>):</strong> Metodo de conjunto (<em>ensemble</em>) que constroi arvores sequenciais para corrigir os erros residuais das iteracoes anteriores.</li>
</ol>
<p style="font-family: 'Segoe UI', Arial, sans-serif; color: #334155; font-size: 14px; line-height: 1.6;">
    Como boa pratica metodologica, separamos os dados com divisao estratificada (75% treino e 25% teste, <code>random_state=42</code>) e aplicamos a padronizacao (<code>StandardScaler</code>) dentro de um <code>Pipeline</code> ajustado apenas sobre os dados de treinamento, evitando qualquer vazamento de dados (<em>data leakage</em>).
</p>

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve,
    confusion_matrix,
    classification_report
)

# Separacao dos atributos e da variavel-alvo originais
X_raw = df_uci.drop('num', axis=1)
y_raw = df_uci['num']

# Divisao estratificada (75% treino, 25% teste)
X_tr, X_te, y_tr, y_te = train_test_split(
    X_raw, y_raw, test_size=0.25, stratify=y_raw, random_state=42
)

# Definicao dos tres novos classificadores em Pipelines seguros contra data leakage
modelos_novos = {
    'Regressao Logistica': Pipeline([
        ('scaler', StandardScaler()),
        ('clf', LogisticRegression(max_iter=1000, random_state=42))
    ]),
    'SVM (Kernel RBF)': Pipeline([
        ('scaler', StandardScaler()),
        ('clf', SVC(kernel='rbf', C=1.0, probability=True, random_state=42))
    ]),
    'Gradient Boosting': Pipeline([
        ('scaler', StandardScaler()),
        ('clf', GradientBoostingClassifier(n_estimators=100, learning_rate=0.05, max_depth=3, random_state=42))
    ])
}

resultados = []
predicoes = {}
probabilidades = {}

for nome, pipeline in modelos_novos.items():
    pipeline.fit(X_tr, y_tr)
    y_hat = pipeline.predict(X_te)
    y_prob = pipeline.predict_proba(X_te)[:, 1]
    
    predicoes[nome] = y_hat
    probabilidades[nome] = y_prob
    
    resultados.append({
        'Modelo': nome,
        'Acuracia': round(accuracy_score(y_te, y_hat), 4),
        'Precision (Classe 1)': round(precision_score(y_te, y_hat), 4),
        'Recall (Classe 1)': round(recall_score(y_te, y_hat), 4),
        'F1-Score (Classe 1)': round(f1_score(y_te, y_hat), 4),
        'ROC-AUC': round(roc_auc_score(y_te, y_prob), 4)
    })
    
    print(f'=== Relatorio de Classificacao: {nome} ===')
    print(classification_report(y_te, y_hat, target_names=['Sem Doenca (0)', 'Com Doenca (1)']))

df_resultados = pd.DataFrame(resultados).sort_values(by='F1-Score (Classe 1)', ascending=False).reset_index(drop=True)
print('=== Quadro Comparativo de Desempenho na Base de Teste ===')
display(df_resultados)
```

```python
# Matrizes de Confusao comparativas para os 3 novos classificadores
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (nome, y_hat) in zip(axes, predicoes.items()):
    cm = confusion_matrix(y_te, y_hat)
    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        cbar=False,
        xticklabels=['Prev: Sem (0)', 'Prev: Com (1)'],
        yticklabels=['Real: Sem (0)', 'Real: Com (1)'],
        ax=ax
    )
    ax.set_title(f'Matriz de Confusao - {nome}', fontsize=11, fontweight='bold', color='#0f172a')

plt.tight_layout()
plt.show()

# Curva ROC comparativa
fig, ax = plt.subplots(figsize=(8, 5))
cores_roc = {'Regressao Logistica': '#2563eb', 'SVM (Kernel RBF)': '#0f172a', 'Gradient Boosting': '#16a34a'}

for nome, y_prob in probabilidades.items():
    fpr, tpr, _ = roc_curve(y_te, y_prob)
    auc_val = roc_auc_score(y_te, y_prob)
    ax.plot(fpr, tpr, label=f'{nome} (AUC = {auc_val:.4f})', linewidth=2, color=cores_roc[nome])

ax.plot([0, 1], [0, 1], linestyle='--', color='#94a3b8', label='Classificador Aleatorio (AUC = 0.50)')
ax.set_title('Curvas ROC Comparativas na Base de Teste', fontsize=12, fontweight='bold', color='#0f172a')
ax.set_xlabel('Taxa de Falsos Positivos (1 - Especificidade)', fontsize=10)
ax.set_ylabel('Taxa de Verdadeiros Positivos (Recall / Sensibilidade)', fontsize=10)
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```
